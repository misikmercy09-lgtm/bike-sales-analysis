```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel(r"C:\Users\User\Desktop\A\Bike sale analysis.xlsx")
df.head()
```

```python
#overview of the dataset
df.info()
```

```python
#looking at the statistics of the dataset
df.describe()
```

```python
#checking for missing values
df.isnull().sum()
```

```python
#checking for duplicates
df.duplicated().sum()
```

```python
#dropping the duplicates
df=df.drop_duplicates()
```

```python
#checking if the duplicates have been dropped
df.duplicated().sum()
```

```python
# Set style
sns.set(style='whitegrid')

# Gender Distribution
sns.countplot(data=df, x='Gender')
plt.title('Customer Gender Distribution')
plt.show()
```

```python
sns.histplot(data=df, x='Age', bins=10, kde=True, color='skyblue')
plt.title('Age Distribution of Customers')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.boxplot(data=df, x='Purchased Bike', y='Income')
plt.title('Income Distribution by Bike Purchase')
plt.xlabel('Purchased Bike')
plt.ylabel('Income')
plt.show()
```

```python
sns.countplot(data=df, x='Commute Distance', hue='Purchased Bike', palette='Set1')
plt.title('Commute Distance vs. Bike Purchase')
plt.xticks(rotation=45)
plt.xlabel('Commute Distance')
plt.ylabel('Count')
plt.show()
```

```python
sns.countplot(data=df, x='Education', hue='Purchased Bike', palette='coolwarm')
plt.title('Education vs. Bike Purchase')
plt.xticks(rotation=45)
plt.xlabel('Education Level')
plt.ylabel('Count')
plt.show()
```

```python
sns.countplot(data=df, x='Marital Status', hue='Purchased Bike', palette='mako')
plt.title('Marital Status vs. Bike Purchase')
plt.xlabel('Marital Status')
plt.ylabel('Count')
plt.show()
```

```python
df['Occupation'].value_counts().plot(kind='bar', color='purple')
plt.title('Occupation Distribution')
plt.xlabel('Occupation')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
